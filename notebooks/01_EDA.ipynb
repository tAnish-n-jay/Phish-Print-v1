```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_style("dark")
print('Libraries loaded successfully!')
```

```python
df = pd.read_csv('../data/dataset.csv', index_col=0)
print(f'Shape: {df.shape}')
df.head()
```

```python
df['status'] = df['status'].map({'legitimate': 0, 'phishing': 1})
print(df['status'].unique())
```

```python
print('Column Names: ')
print(df.columns.tolist())
print('\nData Types: ')
print(df.dtypes.value_counts())
```

```python
missing_data = df.isnull().sum().sum()
print(f"Total missing values: {missing_data}")
```

```python
print(df['status'].value_counts())
print("Percentage: ")
print(df['status'].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='status', data = df, palette='Set2')
plt.title('Class Distribution - Phishing v/s Legitimate')
plt.xticks([0, 1], ['Legitimate', 'Phishing'])
plt.savefig('../results/class_distribution.png')
plt.show()
```

```python
df.describe()
```

```python
plt.figure(figsize=(12, 12))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap='coolwarm', center = 0, linewidths = 0.5, annot = False)
plt.title('Feature Correlation Heatmap')
plt.savefig('../results/correlation_heatmap.png')
plt.show()
```

```python
X= df.drop('status', axis = 1)
Y = df['status']
print(f'Features shape: {X.shape}')
print(f'Target shape: {Y.shape}')
```

```python
features = ['length_url', 'nb_dots', 'nb_hyphens', 
            'nb_subdomains', 'nb_at', 'nb_slash']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, feature in enumerate(features):
    axes[i].hist(df[df['status'] == 0][feature], 
                 alpha=0.5, label='Legitimate', color='green', bins=30)
    axes[i].hist(df[df['status'] == 1][feature],
                 alpha=0.5, label='Phishing', color='red', bins=30)
    axes[i].set_title(feature)
    axes[i].legend()

plt.suptitle('Feature Distribution — Phishing vs Legitimate')
plt.tight_layout()
plt.savefig('../results/feature_distributions.png')
plt.show()
```

```python
target_corr = df.corr()['status'].abs().sort_values(ascending=False)
print("Top 15 features most correlated with target: ")
print(target_corr[1:16])
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='status', y = 'length_url',data = df, palette='Set2')
plt.xticks([0, 1], ['Legitimate', 'Phishing'])
plt.title('URL Length - Phishing vs Legitimate')
plt.savefig('../results/url_length_boxplot.png')
plt.show()
```
